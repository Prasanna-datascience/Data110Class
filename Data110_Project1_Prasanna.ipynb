```python
import pandas as pd
import numpy as np

# ---------------------------
# Load data
# ---------------------------
url = "https://raw.githubusercontent.com/Prasanna-datascience/Data110Class/main/diabetes.prev.csv"

df = pd.read_csv(url)

# ---------------------------
# Create average diabetes rate
# ---------------------------
df["avg_diabetes"] = (
    df["percent.men.diabetes"] +
    df["percent.women.diabetes"]
) / 2

# ---------------------------
# Create average obesity rate
# ---------------------------
df["avg_obesity"] = (
    df["percent.men.obese"] +
    df["percent.women.obese"]
) / 2

# Verify
print(df.head())
print(df.columns.tolist())
```

```python
import pandas as pd

url = "https://raw.githubusercontent.com/Prasanna-datascience/Data110Class/main/diabetes.prev.csv"
df = pd.read_csv(url)

print("Rows, Columns:", df.shape)
print("\nColumns:")
print(df.columns.tolist())

print("\nFirst 5 rows:")
print(df.head())
```

```python
import matplotlib.pyplot as plt

# Ensure the 'avg_diabetes' column exists in df
if "avg_diabetes" not in df.columns:
    df["avg_diabetes"] = (
        df["percent.men.diabetes"] +
        df["percent.women.diabetes"]
    ) / 2

# Histogram of average diabetes prevalence
plt.figure(figsize=(8, 5))

plt.hist(
    df["avg_diabetes"],
    bins=20,
    color="steelblue",
    edgecolor="black"
)

plt.title("Distribution of Average Diabetes Prevalence")
plt.xlabel("Average Diabetes Prevalence (%)")
plt.ylabel("Number of Counties")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(
    df["avg_obesity"],
    df["avg_diabetes"],
    alpha=0.5,
    color="darkgreen",
    label="Counties"
)

plt.title("Obesity vs Diabetes Prevalence")
plt.xlabel("Average Obesity (%)")
plt.ylabel("Average Diabetes (%)")
plt.legend()

plt.show()
```

```python
state_avg = (
    df.groupby("State")["avg_diabetes"]
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(10,5))

state_avg.plot(
    kind="bar",
    color="darkred"
)

plt.title("Top 10 States by Average Diabetes Prevalence")
plt.xlabel("State")
plt.ylabel("Average Diabetes (%)")

plt.tight_layout()
plt.show()
```
